```python
df_resultado = spark.sql("""
    SELECT 
        q.question_id,
        q.question_text,
        a.answer_text,
        SIZE(a.retrieved_chunks) AS total_chunks_used,
        a.llm_model,
        a.created_at
    FROM gold_questions q
    JOIN gold_answers a ON q.question_id = a.question_id
    ORDER BY a.created_at DESC
""")

# En Fabric se pasa el DataFrame como parámetro a la función global display()
display(df_resultado)
```
